```python
import cv2, os

# ====== CONFIGURE YOUR PATHS ======
video_path = "/content/drive/MyDrive/video1.mp4"   # your source video
frames_dir = "/content/drive/MyDrive/frames_OSX"        # where to save frames
fps_target = 30                                         # target fps

# ====== PREPARE OUTPUT FOLDER ======
os.makedirs(frames_dir, exist_ok=True)

# ====== READ VIDEO ======
cap = cv2.VideoCapture(video_path)
orig_fps = cap.get(cv2.CAP_PROP_FPS)
interval = int(orig_fps / fps_target) if orig_fps > 0 else 1

print(f"Extracting from: {video_path}")
print(f"Original FPS: {orig_fps:.2f} → Target FPS: {fps_target} (interval={interval})")

# ====== LOOP AND SAVE FRAMES ======
i, saved = 0, 0
while True:
    ret, frame = cap.read()
    if not ret:
        break
    if i % interval == 0:
        frame_name = f"frame_{saved:05d}.jpg"
        cv2.imwrite(os.path.join(frames_dir, frame_name), frame)
        saved += 1
    i += 1

cap.release()
print(f"Extracted {saved} frames to {frames_dir}/")
```

```python
import os
import cv2
import numpy as np
import imageio

FRAMES_FOLDER = "/content/drive/MyDrive/rendered_frames"   # where your PNG frames are saved
OUTPUT_VIDEO  = "/content/drive/MyDrive/original_output.mp4"
FPS = 30

# 1. Collect frame paths
frames = sorted([f for f in os.listdir(FRAMES_FOLDER) if f.endswith(".png")])
print("[INFO] Total frames:", len(frames))

# 2. Find maximum width and height across all frames
sizes = []
for f in frames:
    img = cv2.imread(os.path.join(FRAMES_FOLDER, f))
    h, w = img.shape[:2]
    sizes.append((h, w))

max_h = max(h for h, w in sizes)
max_w = max(w for h, w in sizes)

print("[INFO] Max frame size:", max_w, "x", max_h)

# 3. Create writer
writer = imageio.get_writer(OUTPUT_VIDEO, fps=FPS)

# 4. Write padded frames
for idx, f in enumerate(frames):
    img = cv2.imread(os.path.join(FRAMES_FOLDER, f))
    h, w = img.shape[:2]

    # Create padded frame (same background she used)
    padded = np.full((max_h, max_w, 3), 245, dtype=np.uint8)
    padded[:h, :w] = img  # top-left placement (like her raw frames)

    writer.append_data(padded)

    if idx % 50 == 0:
        print(f"[INFO] Added frame {idx+1}/{len(frames)}")

writer.close()

print("[DONE] Video saved:", OUTPUT_VIDEO)
```

```python
import numpy as np
from scipy.spatial.transform import Rotation as R
import matplotlib.pyplot as plt

npz_path = "/content/drive/MyDrive/Filtered_GV/frame_00543_params.npz"  # replace with your NPZ path
person_id = 0

Z = np.load(npz_path, allow_pickle=True)
p = f"person_{person_id}_smplx_"

# Extract parameters
root_rot = Z[p+"root_pose"].reshape(3)
body_pose = Z[p+"body_pose"].reshape(-1)
lhand_pose = Z[p+"lhand_pose"].reshape(-1)
rhand_pose = Z[p+"rhand_pose"].reshape(-1)
jaw_pose = Z[p+"jaw_pose"].reshape(3)
betas = Z[p+"shape"].reshape(-1)
expr = Z[p+"expr"].reshape(-1)
cam_trans = Z.get(f"person_{person_id}_cam_trans", np.zeros(3))
focal = Z.get(f"person_{person_id}_focal", [5000.0, 5000.0])
princpt = Z.get(f"person_{person_id}_princpt", [640.0, 360.0])

# Print numeric values
print("Root rotation (axis-angle):", root_rot)
print("Body pose:", body_pose)
print("Left hand pose:", lhand_pose)
print("Right hand pose:", rhand_pose)
print("Jaw pose:", jaw_pose)
print("Shape (betas):", betas)
print("Expression:", expr)
print("Camera translation:", cam_trans)
print("Focal:", focal)
print("Principal point:", princpt)

# Convert root rotation to Euler for easier visualization
rot_deg = R.from_rotvec(root_rot).as_euler('xyz', degrees=True)
print("Root rotation (Euler angles in deg):", rot_deg)

# Optional: visualize pose magnitudes
plt.figure(figsize=(8,4))
plt.plot(body_pose, label="Body pose")
plt.plot(lhand_pose, label="Left hand")
plt.plot(rhand_pose, label="Right hand")
plt.plot(jaw_pose, label="Jaw")
plt.title("Parameter values in this NPZ")
plt.legend()
plt.show()
```

```python
def gaussian_volterra(data, sigma=3, poly=2):
    data = data.astype(np.float32)

    # Add tiny noise so the filter does not collapse
    data = data + np.random.normal(0, 1e-6, data.shape)

    g = gaussian_filter1d(data, sigma=sigma, axis=0)
    t = np.arange(len(data))
    trend = np.polyval(np.polyfit(t, g.mean(axis=1), poly), t).reshape(-1,1)

    return g - g.mean(axis=1, keepdims=True) + trend
```

```python
import cv2, os

def build_video(frames, out):
    files = sorted([f for f in os.listdir(frames) if f.endswith(".jpg")])
    if len(files)==0:
        print("no frames:", frames); return
    img = cv2.imread(f"{frames}/{files[0]}")
    H,W = img.shape[:2]
    v = cv2.VideoWriter(out, cv2.VideoWriter_fourcc(*"mp4v"), 30, (W,H))
    for f in files:
        v.write(cv2.imread(f"{frames}/{f}"))
    v.release()

build_video("/content/drive/MyDrive/gaussian_render/Original_frames",
            "/content/drive/MyDrive/gaussian_render/original.mp4")

build_video("/content/drive/MyDrive/gaussian_render/GV_frames",
            "/content/drive/MyDrive/gaussian_render/smoothed_GV.mp4")

build_video("/content/drive/MyDrive/gaussian_render/SG_frames",
            "/content/drive/MyDrive/gaussian_render/smoothed_SG.mp4")
```
